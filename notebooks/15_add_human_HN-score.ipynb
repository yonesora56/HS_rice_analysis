- Combine rice-human unique LS-HS hit pair and human HN-score previously reported.
- human HN-score list is available from figshare

```python
import polars as pl
```

&nbsp;

## human HN-score

```python
# Downloaded from figshare
# https://doi.org/10.6084/m9.figshare.23944935
HNscore_human = pl.read_csv(
    "../Data/24_meta-analysis_human_mouse/HN-score_human_all.csv",
    separator=","
).rename(
    {
        "GeneName": "hgnc_symbol_id",
        "HN5": "human HN-score",
        "up5": "human up5 pair",
        "dn5": "human down5 pair",
        "unchange5": "human unchange5 pair",
        "all": "human all pair"
    }
)

display(HNscore_human)
```

```python
# In a previous meta-analysis of public gene expression data related to human heat stress, 
# the top 500 genes identified as belonging to the upregulated gene group based on HN-score
# DOI: 10.3390/ijms241713444
HNscore_human_top_500 = HNscore_human.head(500)

top_500_genes = HNscore_human_top_500["hgnc_symbol_id"]

display(HNscore_human_top_500)
```

```python
HNscore_human_bottom_500 = HNscore_human.tail(500)

bottom_500_genes = HNscore_human_bottom_500["hgnc_symbol_id"]

display(HNscore_human_bottom_500)
```

## rice upregulated gene group

```python
foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_250831 = pl.read_csv(
    "../Data/20_foldseek_result_validate_rice_up/rice_up/250831/foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_250831.tsv",
    separator="\t"
).join(
    HNscore_human,
    on="hgnc_symbol_id",
    how="left"
).with_columns(
    (pl.col("hgnc_symbol_id").is_in(top_500_genes).alias("is_in_human_top500")),
    (pl.col("hgnc_symbol_id").is_in(bottom_500_genes).alias("is_in_human_bottom500"))
)

foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_250831.write_csv(
    "../Data/20_foldseek_result_validate_rice_up/rice_up/rice_up_supp9_add_human_HNscore/foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_250831_add_human_HNscore.tsv",
    separator="\t"
)

display(foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_250831)
```

```python
foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter1_250831 = pl.read_csv(
    "../Data/20_foldseek_result_validate_rice_up/rice_up/250831/foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter1_250831.tsv",
    separator="\t"
).join(
    HNscore_human,
    on="hgnc_symbol_id",
    how="left"
).with_columns(
    (pl.col("hgnc_symbol_id").is_in(top_500_genes).alias("is_in_human_top500")),
    (pl.col("hgnc_symbol_id").is_in(bottom_500_genes).alias("is_in_human_bottom500"))
)

foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter1_250831.write_csv(
    "../Data/20_foldseek_result_validate_rice_up/rice_up/rice_up_supp9_add_human_HNscore/foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter1_250831_add_human_HNscore.tsv",
    separator="\t"
)

display(foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter1_250831)
```

```python
foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter4_250831 = pl.read_csv(
    "../Data/20_foldseek_result_validate_rice_up/rice_up/250831/foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter4_250831.tsv",
    separator="\t"
).join(
    HNscore_human,
    on="hgnc_symbol_id",
    how="left"
).with_columns(
    (pl.col("hgnc_symbol_id").is_in(top_500_genes).alias("is_in_human_top500")),
    (pl.col("hgnc_symbol_id").is_in(bottom_500_genes).alias("is_in_human_bottom500"))
)

foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter4_250831.write_csv(
    "../Data/20_foldseek_result_validate_rice_up/rice_up/rice_up_supp9_add_human_HNscore/foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter4_250831_add_human_HNscore.tsv",
    separator="\t"
)

print(foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter4_250831.select("hit label").unique())
display(foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter4_250831)
```

```python
rice_up_human_up = foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter4_250831.filter(
    pl.col("is_in_human_top500") == True
)

rice_up_human_up.write_csv(
    "../Data/20_foldseek_result_validate_rice_up/rice_up/rice_up_supp9_add_human_HNscore/foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter4_250831_add_human_HNscore_top500.tsv",
    separator="\t"
)

pl.Config.set_tbl_rows(14)
print(rice_up_human_up.select("hit label").unique())
display(rice_up_human_up)
```
