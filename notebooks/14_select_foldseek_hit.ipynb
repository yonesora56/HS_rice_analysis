## rice upregulated gene group "unique LS-HS" hit pair analysis

```python
rice_up_unique_ls_hs_hit_pair = pl.read_csv(
    "../Data/20_foldseek_result_validate_rice_up/rice_up/250831/foldseek_result_rice_up_domain_panhomology_info_filtercondtion3_filter4_250831.tsv",
    separator="\t"
)

display(rice_up_unique_ls_hs_hit_pair)
```

&nbsp;

## InterPro ID & Ensembl pan-homology info & no enrichment result

```python
rice_up_unique_ls_hs_hit_pair_ip_ensembl = rice_up_unique_ls_hs_hit_pair.filter(
    (pl.col("enrichment_result") == False) & # This is not encluded in the enrichment analysis
    (pl.col("type").is_not_null()) # This mentions that the Ensembl pan-homology information is available
)

# candidate hit pair 1 Os03g0830500 (HN-score: 61)
Os03g0830500 = rice_up_unique_ls_hs_hit_pair.filter(
    (pl.col("From") == "Os03g0830500")
)

# candidate hit pair 2 Os08g0442200 (HN-score: 48) ← keep this pair
Os08g0442200 = rice_up_unique_ls_hs_hit_pair.filter(
    (pl.col("From") == "Os08g0442200")
)

display(Os03g0830500)
display(Os08g0442200)
display(rice_up_unique_ls_hs_hit_pair_ip_ensembl)
```

## InterPro ID & no enrichment result

### Replace "A0A0P0Y9Z0" to another entry in rice_up_unique_ls_hs_hit_pair

```python
rice_up_unique_ls_hs_hit_pair_replace = rice_up_unique_ls_hs_hit_pair.filter(
    (
        (pl.col("enrichment_result") == False) & # This is not encluded in the enrichment analysis
        (pl.col("InterPro ID intersection tag") == "have common InterPro ID") &
        (pl.col("type").is_null().all().over("From")) # All rows for this "From" gene have null type (no Ensembl pan-homology info at gene level)
    ) & 
    (
        (pl.col("From") != "Os12g0478200") & # pair accession "A0A0P0Y9Z0" (highest average lDDT) is obsolete (2025-12-08)
        (pl.col("From") != "Os01g0860601") & # pair accession "Q5N7C3" (highest average lDDT) is obsolete (2025-12-08)
        (pl.col("From") != "Os02g0711300") & # pair accession "Q0DY72" (highest average lDDT) is obsolete (2025-12-08)
        (pl.col("From") != "Os04g0185600") &  # pair accession "Q7X7V2" (highest average lDDT) is obsolete (2025-12-08)
        (pl.col("From") != "Os02g0649300") &  # human accession "P52954" is obsolete (2025-12-08)
        (pl.col("From") != "Os03g0820300")   # pLDDT is not high overall, so filtering (e.g. E5RH04)
    )
)

# candidate hit pair 1 Os01g0348900 (HN-score: 53)
Os01g0348900 = rice_up_unique_ls_hs_hit_pair.filter(
    (pl.col("From") == "Os01g0348900")
)

# candidate hit pair 2 Os09g0444900 (HN-score: 45) ← keep this pair
Os09g0444900 = rice_up_unique_ls_hs_hit_pair.filter(
    (pl.col("From") == "Os09g0444900")
)

display(Os01g0348900)
display(Os09g0444900)
display(rice_up_unique_ls_hs_hit_pair_replace)
```

## no enrichment result

```python
rice_up_unique_ls_hs_hit_pair_no_enrichment = rice_up_unique_ls_hs_hit_pair.filter(
    (
        (pl.col("enrichment_result") == False) &
        (pl.col("InterPro ID intersection tag") != "have common InterPro ID")
    )
)

# candidate hit pair 1 Os03g0817100 (HN-score: 54) ← keep this pair
Os03g0817100 = rice_up_unique_ls_hs_hit_pair_no_enrichment.filter(
    (pl.col("From") == "Os03g0817100")
).sort(
    by="lddt",
    descending=True
)

display(Os03g0817100)
display(rice_up_unique_ls_hs_hit_pair_no_enrichment)
```
