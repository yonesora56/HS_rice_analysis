## Check Status between AFDB version 4 and 6

- https://www.ebi.ac.uk/pdbe/news/alphafold-database-release-notes

```text
Key statistics

AlphaFold DB v4: 214,683,829 predicted protein structures
AlphaFold DB v6: 241,070,489 predicted protein structures  (including 40,054 isoforms)
Differences in the number of proteins since v4:

Added: 65,711,653
Removed: 39,324,993
Changed: 276,539
Unchanged: 175,082,297
Interpretation:

“Added” and “Removed” follow upstream UniProt: new proteins predicted, and records archived upstream are retired. “Added” includes isoforms too.
“Changed” refers to amino-acid sequence changes (not metadata).
“Unchanged” entries received metadata updates to reflect current UniProt information and are relabelled v6. The underlying model coordinates have not been updated, but carried over from v4.
```

````python
afdb_accession_status = pl.read_ndjson(
    "../Data/23_AFDB/diffs.ndjson" # Downloaded from https://ftp.ebi.ac.uk/pub/databases/alphafold/diffs.ndjson.gz
).filter(
    pl.col("status") != "ADDED"
)

display(afdb_accession_status)
````
